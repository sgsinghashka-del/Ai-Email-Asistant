```python
# Import the required libraries
from google import genai
from getpass import getpass


# Enter the Gemini API key securely
api_key = getpass("Enter your Gemini API key: ")


# Connect Python with Gemini
client = genai.Client(api_key=api_key)


# Take email input from the user
email_text = input("\nEnter your email:\n")


# Create a prompt for Gemini
prompt = f"""
Improve the following email.

Instructions:
- Correct grammar and spelling.
- Improve sentence structure.
- Make the email clear and professional.
- Keep the original meaning.
- Return only the improved email.

Email:
{email_text}
"""


# Send the prompt to Gemini
response = client.models.generate_content(
    model="gemini-3.6-flash",
    contents=prompt
)


# Display the improved email
print("\nImproved Email:\n")
print(response.text)
```

```python
from google import genai
from getpass import getpass

# Enter the Gemini API key
api_key = getpass("Enter your Gemini API key: ")

# Connect Python with Gemini
client = genai.Client(api_key=api_key)

# Take booking details from the user
guest_name = input("Enter guest name: ")
check_in = input("Enter check-in date: ")
check_out = input("Enter check-out date: ")
number_of_guests = input("Enter number of guests: ")
room_type = input(
    "Enter room type (Deluxe, Premium or Family): "
)
special_request = input(
    "Enter special request or write None: "
)

# Create the prompt
prompt = f"""
You are a hotel booking assistant for Sunrise Stay Hotel, Delhi.

Prepare a clear and professional booking request response.

Booking Details:
Guest Name: {guest_name}
Check-in Date: {check_in}
Check-out Date: {check_out}
Number of Guests: {number_of_guests}
Room Type: {room_type}
Special Request: {special_request}

Instructions:
- Begin with a polite greeting.
- Mention all booking details correctly.
- Use clear and professional language.
- Mention the special request.
- Do not say that the booking is confirmed.
- Mention that the booking is subject to room availability.
- Return only the booking response.
- Don't use '**' in the response.
"""

# Send the prompt to Gemini
response = client.models.generate_content(
    model="gemini-3.6-flash",
    contents=prompt
)

# Display the generated response
print("\nHotel Booking Response:\n")
print(response.text)
```
